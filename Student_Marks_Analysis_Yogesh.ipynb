```python
# Student Marks Analysis

marks = [85, 72, 90, 65, 78, 88, 56, 95]

# Using built-in functions
total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)

print("Student Marks:", marks)
print("Total Marks:", total)
print("Average Marks:", average)
print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
```

```python
# Manual calculation using loops

total_manual = 0

for mark in marks:
    total_manual += mark

average_manual = total_manual / len(marks)

highest_manual = marks[0]
lowest_manual = marks[0]

for mark in marks:
    if mark > highest_manual:
        highest_manual = mark
    if mark < lowest_manual:
        lowest_manual = mark

print("Manual Total:", total_manual)
print("Manual Average:", average_manual)
print("Manual Highest:", highest_manual)
print("Manual Lowest:", lowest_manual)
```

```python

# Passing mark analysis

passing_mark = 40

passed_students = 0
failed_students = 0

for mark in marks:
    if mark >= passing_mark:
        passed_students += 1
    else:
        failed_students += 1

print("Passing Mark:", passing_mark)
print("Number of Passed Students:", passed_students)
print("Number of Failed Students:", failed_students)
```
